```python
import pandas as pd

url = "https://raw.githubusercontent.com/CunyLaguardiaDataAnalytics/datasets/master/2014-15_To_2016-17_School-_Level_NYC_Regents_Report_For_All_Variables.csv"
df = pd.read_csv(url)

df.head()
```

```python
df.shape
```

```python
school =df[df["School Name"]== "P.S. 034 Franklin D. Roosevelt"]
school["Regents Exam"].unique()
```

```python
school_exam = school[school["Regents Exam"] == "Living Enviroment"]

school_exam[["School Name", "Year", "Total Tested", "Mean Score"]]
```

```python
df[df["School DBN"]=="01M034"]
```

```python
df["Mean Score"] = pd.to_numeric(df["Mean Score"], errors="coerce")
df=df.dropna(subset=["Mean Score"])
```

```python
df=df.copy()
```

```python
df["Mean Score"].describe()
```

```python
exam = df[df["Regents Exam"]=="Living Enviroment"]
school_avg = exam[exam["School DBN"]=="01M034"]["Mean Score"].mean()
nyc_avg = exam["Mean Score"].mean()
print("School Average:", round(school_avg,2))
print("NYC Average:", round(nyc_avg,2))
```

```python
df["Regents Exam"].unique()
```

```python
exam = df[df["Regents Exam"]=="Living Environment"]

exam[["School DBN","Regents Exam","Mean Score"]].head()
```

```python
school_avg = exam[exam["School DBN"]=="01M034"]["Mean Score"].mean()
nyc_avg = exam["Mean Score"].mean()
print("School Average:", round(school_avg,2))
print("NYC Average:", round(nyc_avg,2))
```

```python
import matplotlib.pyplot as plt

plt.bar(["P.S 034", "NYC"], [school_avg, nyc_avg])

plt.title("Living Environment Exam Scores")
plt.ylabel("Average Score")

plt.show()
```

For this project I compared P.S 034 Franklin D Roosevelt's Living Environment regents scores within the NYC average.

The school average score was 76.75 when compared to the NYC average of 69.06 which shows that the school scored higher than the NYC average.
